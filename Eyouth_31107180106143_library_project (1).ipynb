```python
import sqlite3
import pandas as pd
import json
# Connect to the database
conn = sqlite3.connect("library.db")

# Read the members table
members = pd.read_sql_query(
    "SELECT * FROM members",
    conn
)

# Read the checkouts table
checkouts = pd.read_sql_query(
    "SELECT * FROM checkouts",
    conn
)
# Read the books table
books = pd.read_sql_query(
    "SELECT * FROM books",
    conn
)



print("Members:")
print(members.head())

print("\nCheckouts:")
print(checkouts.head())

print("\nBooks:")
print(books.head())





```

```python
#seeing all members checkouts
query1 = """
SELECT
    members.member_id,
    members.first_name,
    members.last_name,
    COUNT(checkouts.book_id) AS total_checkouts
FROM members
LEFT JOIN checkouts
    ON members.member_id = checkouts.member_id
GROUP BY
    members.member_id,
    members.first_name,
    members.last_name
ORDER BY total_checkouts DESC;
"""

result1 = pd.read_sql_query(query1, conn)

print(result1)
```

```python
#the books whose author name starts with A
q2 = "SELECT * FROM books WHERE author LIKE 'A%';"
result2 = pd.read_sql_query(q2, conn)
print(result2)
```

```python
#the most reading books
q3 = """ SELECT book_id, COUNT(*) AS checkout_count FROM checkouts
GROUP BY book_id ORDER BY checkout_count DESC LIMIT 5; """
result3= pd.read_sql_query(q3, conn)
print(result3)
```

```python
#the most active readers
q4 = """
SELECT
    members.member_id,
    members.first_name,
    members.last_name,
    COUNT(checkouts.book_id) AS books_borrowed
FROM members
JOIN checkouts
    ON members.member_id = checkouts.member_id
GROUP BY
    members.member_id,
    members.first_name,
    members.last_name
ORDER BY books_borrowed DESC
LIMIT 10;
"""

result4 = pd.read_sql_query(q4, conn)

print(result4)
```

```python
#the recent checkouts
q5 = """
SELECT
    checkouts.checkout_id,
    checkouts.member_id,
    checkouts.book_id,
    checkouts.checkout_date,
    checkouts.return_date,
    members.neighborhood
FROM checkouts
JOIN members
    ON checkouts.member_id = members.member_id
WHERE LOWER(members.neighborhood) = 'maadi'
ORDER BY checkouts.checkout_date DESC
LIMIT 5 OFFSET 10;
"""

result5 = pd.read_sql_query(q5, conn)

print(result5)
```

```python
#connecting checkouts table with members table
data = pd.merge(
    checkouts,
    members,
    on="member_id",
    how="left"
)

print(data.shape)
```

```python
#connecting the books table with the other two tables
db = pd.merge(
    data,
    books,
    on="book_id",
    how="left"
)

print(db.shape)
print(db.head())
```

```python
books = pd.read_json("books.json")
db_json = pd.merge(
    db,
    books,
    on="book_id",
    how="left"
)
print(db_json.shape)
print(db_json.head())
```

```python
checkout = pd.read_html("checkout.html")[0]

print(checkout.shape)
print(checkout.head())
```

```python
all_data = pd.concat(
    [db_json, checkout],
    ignore_index=True
)

print(all_data.shape)
print(all_data.head())
```

```python
print("Final shape:")
print(all_data.shape)
print("\nColumns:")
print(all_data.columns)
print("\nFirst 5 rows:")
print(all_data.head())
```

```python
all_data.to_csv(
    "task1_combined_data.csv",
    index=False
)

print("File saved successfully!")
```

### Web Scraping vs API

**Web Scraping:** Collects data from the HTML structure of a webpage. It can break if the webpage design changes.

**API:** Provides data directly in a structured format such as JSON. It is more stable, reliable, and easier to process.

**Importance in our project:** Using an API would make collecting and combining library data easier and more reliable, with fewer errors than depending on HTML structure.
